### Importación de datos



```python
import pandas as pd

url1 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda1 = pd.read_csv(url1)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)

tiendas = [tienda1,tienda2,tienda3,tienda4]

tiendas[0].head()

```

```python
tiendas_concatenadas = pd.concat(tiendas)
```

#1. Análisis de facturación



```python
ingresos = []
for i in range(1,5):
  ingresos.append(tiendas[i-1].Precio.sum())

print(ingresos)
```

```python
import matplotlib.pyplot as plt
```

```python
nombre_tiendas = ['Tienda 1','Tienda 2','Tienda 3','Tienda 4']

fig, ax = plt.subplots(1,2, figsize=(15,5))
# Primer gráfico (barras verticales)
ax[0].bar(nombre_tiendas, ingresos)
ax[0].set_title('Ingresos por tienda')
ax[0].set_xlabel('Tiendas')
ax[0].set_ylabel('Ingresos')
ax[0].bar_label(ax[0].containers[0],fmt="%.2f")

# Segundo gráfico (torta)
ax[1].pie(ingresos,autopct="%1.1f%%")
ax[1].set_title('Porcentaje de ingresos por tienda')
ax[1].legend(nombre_tiendas)


plt.show()

#De los graficos dibujados para los ingresos, el mayor ingreso es provisto por la Tienda 1, seguido de la Tienda 2, Tienda 3 y Tienda 4 en ese orden
```

# 2. Ventas por categoría

```python
#Lista de categorias para todas las tiendas
categorias = tiendas_concatenadas['Categoría del Producto'].unique()
print(categorias)
```

```python
#Series de Pandas con las cantidades por categorias para las tiendas
cantidad_t1 = tienda1['Categoría del Producto'].value_counts()
cantidad_t2 = tienda2['Categoría del Producto'].value_counts()
cantidad_t3 = tienda3['Categoría del Producto'].value_counts()
cantidad_t4 = tienda4['Categoría del Producto'].value_counts()
```

```python
#Numero de ventas por categoria por cada tienda
venta_categorias = pd.DataFrame({
    'Tienda 1': cantidad_t1,
    'Tienda 2': cantidad_t2,
    'Tienda 3': cantidad_t3,
    'Tienda 4': cantidad_t4
}, index = categorias)
venta_categorias.sort_index(inplace=True)

venta_categorias
```

```python
categorias_populares = []
for i in range(4):
  categorias_populares.append(venta_categorias.iloc[:,i].idxmax())

df_categorias_populares = pd.DataFrame({
    'Tiendas': nombre_tiendas,
    'Categorias populares': categorias_populares,
})
df_categorias_populares#se observa que la categoria popular de todas las tiendas es de Muebles
```

```python
#Reorganizamos el dataframe de cantidades por categoria para dibujar un grafico de barras acumuladas
cantidad_t1 = pd.DataFrame({
    'Cantidad': cantidad_t1,
    'Tienda': 'Tienda 1'
})
cantidad_t2 = pd.DataFrame({
    'Cantidad': cantidad_t2,
    'Tienda': 'Tienda 2'
})
cantidad_t3 = pd.DataFrame({
    'Cantidad': cantidad_t3,
    'Tienda': 'Tienda 3'
})
cantidad_t4 = pd.DataFrame({
    'Cantidad': cantidad_t4,
    'Tienda': 'Tienda 4'
})

cantidad_t1.reset_index(inplace=True)
cantidad_t2.reset_index(inplace=True)
cantidad_t3.reset_index(inplace=True)
cantidad_t4.reset_index(inplace=True)

cantidades_por_categoria = pd.concat([cantidad_t1,cantidad_t2,cantidad_t3,cantidad_t4])
cantidades_por_categoria
```

```python
#Con seaborn dibujamos barras agrupadas
import seaborn as sns
sns.barplot(data=cantidades_por_categoria, x='Tienda', y='Cantidad', hue='Categoría del Producto')
plt.show()
```

```python
#Con Plotly dibujamos barras acumuladas observandose que el mayor valor es Muebles seguido de los Electronicos y Juguetes
#que se ve en la base de cada barra agrupada
import plotly.express as px
fig = px.bar(cantidades_por_categoria, x='Tienda', y='Cantidad', color='Categoría del Producto', barmode='stack')
fig.show()
```

```python
categorias.sort()
categorias
```

```python
#Con Matplotlib realizo graficos circulares por categoria para cada tienda

valores_t1 = cantidades_por_categoria.query("Tienda == 'Tienda 1'").sort_values(by='Categoría del Producto', ascending=True)
valores_t2 = cantidades_por_categoria.query("Tienda == 'Tienda 2'").sort_values(by='Categoría del Producto', ascending=True)
valores_t3 = cantidades_por_categoria.query("Tienda == 'Tienda 3'").sort_values(by='Categoría del Producto', ascending=True)
valores_t4 = cantidades_por_categoria.query("Tienda == 'Tienda 4'").sort_values(by='Categoría del Producto', ascending=True)

figy, axy = plt.subplots(2,2, figsize=(8,8))


axy[0,0].pie(x = valores_t1['Cantidad'],autopct= "%.1f%%")
axy[0,0].set_title('Tienda 1')

axy[0,1].pie(x = valores_t2['Cantidad'],autopct= "%.1f%%")
axy[0,1].set_title('Tienda 2')

axy[1,0].pie(x = valores_t3['Cantidad'],autopct= "%.1f%%")
axy[1,0].set_title('Tienda 3')

axy[1,1].pie(x = valores_t4['Cantidad'],autopct= "%.1f%%")
axy[1,1].set_title('Tienda 4')

figy.legend(categorias, ncol = 4, loc = 'lower center',bbox_to_anchor=(0.5, -0.05))
figy.suptitle('Unidades Vendidas por Categoria por Tienda',fontsize=18)

plt.tight_layout()
plt.show()
#En resumen se observa un mayor numero de ventas en unidades en la categoria de Muebles
```

```python
#Juntamos los dataframe con la columna Tienda para el nuevo dataframe
tienda1_act = tienda1.copy()
tienda1_act['Tienda'] = 'Tienda 1'

tienda2_act = tienda2.copy()
tienda2_act['Tienda'] = 'Tienda 2'

tienda3_act = tienda3.copy()
tienda3_act['Tienda'] = 'Tienda 3'

tienda4_act = tienda4.copy()
tienda4_act['Tienda'] = 'Tienda 4'

tiendas_act = pd.concat([tienda1_act,tienda2_act,tienda3_act,tienda4_act])
tiendas_act.reset_index(drop=True, inplace=True)
tiendas_act.tail()
```

```python
#Se obtiene el precio promedio por categoria para validar si se esta vendiendo en su mayoria los productos
#que dan mayor ganancia por tienda
precio_promedio_t1 = round(tiendas_act.groupby('Categoría del Producto')['Precio'].mean().sort_values(ascending=False),2)
precio_promedio_t1
#Se observa que la categoria con mayor precio es la de Electrodomesticos
```

# 3. Calificación promedio de la tienda


```python
#Para el promedio de calificacion por tienda, se observa una conformidad de 4 sobre 5 que seria el 80% de calificacion para todas las tiendas
tiendas_act.groupby('Tienda')['Calificación'].mean()
```

```python
#Sin embargo observamos que el ranking de calificacion es de Tienda1 < Tienda4 < Tienda2 < Tienda3
ranking_calificacion = tiendas_act.groupby('Tienda')['Calificación'].mean().sort_values(ascending=False)
plt.bar(ranking_calificacion.index, ranking_calificacion.values)
plt.xlabel('Tiendas')
plt.ylabel('Calificación promedio')
plt.title('Calificación promedio por tienda')
plt.show()
```

# 4. Productos más y menos vendidos

```python
productos_t1 = tiendas_act.query("Tienda == 'Tienda 1'")['Producto'].value_counts()
productos_t2 = tiendas_act.query("Tienda == 'Tienda 2'")['Producto'].value_counts()
productos_t3 = tiendas_act.query("Tienda == 'Tienda 3'")['Producto'].value_counts()
productos_t4 = tiendas_act.query("Tienda == 'Tienda 4'")['Producto'].value_counts()

productos_mas_vendidos_t1 = productos_t1.head().sort_values(ascending=True)
productos_mas_vendidos_t2 = productos_t2.head().sort_values(ascending=True)
productos_mas_vendidos_t3 = productos_t3.head().sort_values(ascending=True)
productos_mas_vendidos_t4 = productos_t4.head().sort_values(ascending=True)

productos_menos_vendidos_t1 = productos_t1.tail()
productos_menos_vendidos_t2 = productos_t2.tail()
productos_menos_vendidos_t3 = productos_t3.tail()
productos_menos_vendidos_t4 = productos_t4.tail()
```

```python
#Graficando los 05 productos mas vendidos por cada tienda
fig, axx = plt.subplots(2,2, figsize=(30,10))

fig.suptitle('Productos más vendidos por tienda',fontsize=18)

axx[0,0].barh(productos_mas_vendidos_t1.index, productos_mas_vendidos_t1.values)
axx[0,0].set_title('Tienda 1')

axx[0,1].barh(productos_mas_vendidos_t2.index, productos_mas_vendidos_t2.values)
axx[0,1].set_title('Tienda 2')

axx[1,0].barh(productos_mas_vendidos_t3.index, productos_mas_vendidos_t3.values)
axx[1,0].set_title('Tienda 3')

axx[1,1].barh(productos_mas_vendidos_t4.index, productos_mas_vendidos_t4.values)
axx[1,1].set_title('Tienda 4')


plt.show()
```

```python
#Graficando los 05 productos menos vendidos por cada tienda
fig, axx = plt.subplots(2,2, figsize=(30,10))

fig.suptitle('Productos menos vendidos por tienda',fontsize=18)

axx[0,0].barh(productos_menos_vendidos_t1.index, productos_menos_vendidos_t1.values)
axx[0,0].set_title('Tienda 1')

axx[0,1].barh(productos_menos_vendidos_t2.index, productos_menos_vendidos_t2.values)
axx[0,1].set_title('Tienda 2')

axx[1,0].barh(productos_menos_vendidos_t3.index, productos_menos_vendidos_t3.values)
axx[1,0].set_title('Tienda 3')

axx[1,1].barh(productos_menos_vendidos_t4.index, productos_menos_vendidos_t4.values)
axx[1,1].set_title('Tienda 4')


plt.show()
```

# 5. Envío promedio por tienda

```python
#Comparando en grafico de barras
plt.bar(costo_envio_por_tienda['Tienda'], costo_envio_por_tienda['Costo_Envios'])
plt.xlabel('Tiendas')
plt.ylabel('Costo de envío promedio')
plt.title('Costo de envío promedio por tienda')
plt.show()
#Se observa que el costo de envio promedio por tienda es superior para la tienda1
```
